```python
import pandas as pd
import numpy as np

# 1. Load your existing dependent variable file
csti_df = pd.read_csv("nigeria_state_csti_index.csv")

# 2. Define the unified data dictionary container 
macro_data = {
    "State": [
        "ABIA", "ADAMAWA", "AKWA IBOM", "ANAMBRA", "BAUCHI", "BAYELSA", 
        "BENUE", "BORNO", "CROSS RIVER", "DELTA", "EBONYI", "EDO", 
        "EKITI", "ENUGU", "FCT ABUJA", "GOMBE", "IMO", "JIGAWA", 
        "KADUNA", "KANO", "KATSINA", "KEBBI", "KOGI", "KWARA", 
        "LAGOS", "NASARAWA", "NIGER", "OGUN", "ONDO", "OSUN", 
        "OYO", "PLATEAU", "RIVERS", "SOKOTO", "TARABA", "YOBE", "ZAMFARA"
    ],
    "Public_Order_Safety_Budget": [
        0, 0, 0, 0, 0, 0,      # ABIA to BAYELSA
        0, 0, 0, 0, 0, 0,      # BENUE to EDO
        0, 0, 0, 0, 0, 0,      # EKITI to JIGAWA
        0, 0, 0, 0, 0,         # KADUNA to KOGI
        1800000000,            # KWARA (Updated with absolute Naira calculation)
        0, 0, 0, 0, 0, 0,      # LAGOS to OSUN
        0, 0, 0, 0, 0, 0, 0    # OYO to ZAMFARA
    ],
    "State_Population": [
        4260000, 4900000, 6500000, 6360000, 7740000, 2550000, 
        6800000, 6950000, 4470000, 6650000, 3310000, 4850000, 
        3820000, 5130000, 3560000, 3820000, 6350000, 6730000, 
        9590000, 15420000, 9100000, 5190000, 4970000, 3590000, 
        14730000, 2860000, 6510000, 6150000, 5430000, 5520000, 
        9290000, 4700000, 8660000, 5810000, 3590000, 3830000, 5300000 
    ]
}

# Convert the structural framework into a pandas DataFrame
macro_df = pd.DataFrame(macro_data)

# 3. Execute math operations across vectors
macro_df['Budget_i'] = macro_df['Public_Order_Safety_Budget'] / macro_df['State_Population']
macro_df['ln_Budget_i'] = np.log1p(macro_df['Budget_i'])

# 4. Merge datasets and export clean cross-section
final_regression_df = pd.merge(csti_df, macro_df, on="State", how="inner")
final_regression_df.to_csv("master_regression_dataset.csv", index=False)

print("SUCCESS: Complete dataset constructed and saved as 'master_regression_dataset.csv'")
print(final_regression_df[['State', 'CSTI', 'Budget_i', 'ln_Budget_i']].head(25))
```
